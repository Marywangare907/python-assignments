TASK 1: Load and Explore the Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
```

```python
# Load Iris dataset from sklearn
iris = load_iris()
```

```python
# Create a DataFrame with feature data
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
```

```python
# Add species column from target values
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)
```

```python
# Display the first few rows of the DataFrame
print("First 5 rows of the dataset:")
print(df.head())
```

```python
# Check structure and data types
print("\nDataset info:")
print(df.info())
```

```python
# Check for missing values
print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
# Clean dataset: Drop missing values (not necessary here, but included for completeness)
df.dropna(inplace=True)
```

TASK 2: Basic Data Analysis

```python
# Summary statistics
print("\nSummary statistics:")
print(df.describe())
```

```python
# Group by species and calculate the mean of each feature
grouped = df.groupby('species').mean()
print("\nMean values grouped by species:")
print(grouped)
```

Observations:
Setosa has the smallest petal length and width.
Virginica has the largest measurements across most features.
This helps distinguish between the species clearly.

TASK 3: Data Visualization

```python
# Plot 1: Line Chart of Sepal Dimensions Over Index (simulating time)
plt.figure(figsize=(8, 5))
plt.plot(df.index, df['sepal length (cm)'], label='Sepal Length')
plt.plot(df.index, df['sepal width (cm)'], label='Sepal Width')
plt.title("Sepal Dimensions Over Index (Simulated Time)")
plt.xlabel("Index")
plt.ylabel("Measurement (cm)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Plot 2: Bar Chart - Average Petal Length by Species
grouped['petal length (cm)'].plot(kind='bar', color='skyblue')
plt.title("Average Petal Length by Species")
plt.ylabel("Petal Length (cm)")
plt.xlabel("Species")
plt.tight_layout()
plt.show()
```

```python
# Plot 3: Histogram - Distribution of Petal Length
sns.histplot(df['petal length (cm)'], bins=20, kde=True, color='green')
plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length (cm)")
plt.tight_layout()
plt.show()
```

```python
# Plot 4: Scatter Plot - Sepal Length vs Petal Length
sns.scatterplot(data=df, x='sepal length (cm)', y='petal length (cm)', hue='species')
plt.title("Sepal Length vs Petal Length by Species")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.tight_layout()
plt.show()
```
